```python
def is_even(num):
    """
    This function is tells if a given number isodd or even.
    input - any number
    output - odd/even
    """
    if num %2 == 0:
        return 'Even'
    else:
        return 'Odd'
    
```

```python
is_even(20)
```

```python
for i in range(1,11):
    print(i ,": ",is_even(i))
```

```python
# to get the documentation of any function ny using (.__doc__)
print(is_even.__doc__)
```

```python
def power(a=1,b=1):
    return a**b
```

```python
# Positional argument - position are matter's 2 store in a or 3 store in b
power(2,3)
```

```python
# here code will not crash due to default value
power()
```

```python
# Keyword Argument - this is usefun when you have many aregument and you dont remember it . 
power(b=2,a=3)
```

#### Local and global variable

```python
def func_a():
    print('inside the func_a')

def func_b(y):
    print('inside func_b')
    return y
```

```python
print(func_a())
print(5 + func_b(2))
```

```python
def f(y):
    x = 1
    x += 1
    print(x)
    
x = 5
f(x)
print(x)
```

```python
def g(y):
    print(x)
    print(x+1)
    
x = 5
g(x)
print(x)
```

```python
def h(y):
    x += 1

x=5
h(x)
print(x)
```

```python
def h(y):
    global x
    x += 1

x=5
h(x)
print(x)
```

```python
def f(x):
    x = x+1
    print('in f(x): x =',x)
    return x

x = 3
z= f(x)
print('in main program scope: z =',z)
print('in main program scope: x =',x)
```

```python
def f():
    print("Inside f")
    def g():
        print('Inside g')
    g()
    
f()
```

```python
def g(x):
    def h():
        x = 'abc'
    x = x+1
    print('in g(x): x=',x)
    h()
    return x

x = 3
z = g(x)
```

```python
def g(x):
    def h(x):
        x= x+1
        print('in h(x): x =',x)
    x = x+1
    print('in g(x) = ',x)
    h(x)
    return x

x = 3
z= g(x)
print('in main program scope: x = ',x)
print('in main program scope: z = ',z)

```
