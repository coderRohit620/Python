```python
def is_even(num):
    """
    This function is tells if a given number isodd or even.
    input - any number
    output - odd/even
    """
    if num %2 == 0:
        return 'Even'
    else:
        return 'Odd'
    
```

```python
is_even(20)
```

```python
for i in range(1,11):
    print(i ,": ",is_even(i))
```
