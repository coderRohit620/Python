#### Set Operation

```python
s7 = {1,2,3,4,5,6}
s8 = {3,4,5,6,7,8}
```

```python
# Concatination not allowed
s7+s8
```

```python
# multiplication not allowed
s7 * 3
```

```python
for i in s7:
    print(i);
```
