#### Recursion - a function call its self again and again.

```python
def multiply(a,b):
    result = 0
    
    for i in range(b):
        result = result+a
    print(result)
    
multiply(3,4)
```

```python
# do with resursion
def mul(a,b):
    if b==1:
        return a
    else:
        return a+mul(a,b-1)
    
print(mul(3,4))
```

```python
def fact(num):
    if num ==1:
        return 1
    else:
        # 5 = 5 * 4fact
        num  = num * fact(num-1)
        return num
    
print(fact(5))
```

```python
# check palindrome
def palindrome(str):
    if len(str) <= 1:
        print("Palindrome") 
    else:
        if str[0] == str[-1]:
            palindrome(str[1:-1])
        else:
            print("Not Palindrome") 
```

```python
print(palindrome("madam"))
print(palindrome("abba"))
print(palindrome("rohit"))
```

```python
def fib(m):
    if m==0 or m==1:
        return 1
    else:
        return fib(m-1) + fib(m-2)
    
print(fib(12))
```
