#### Recursion - a function call its self again and again.

```python
def multiply(a,b):
    result = 0
    
    for i in range(b):
        result = result+a
    print(result)
    
multiply(3,4)
```

```python
# do with resursion
def mul(a,b):
    if b==1:
        return a
    else:
        return a+mul(a,b-1)
    
print(mul(3,4))
```

```python
def fact(num):
    if num ==1:
        return 1
    else:
        # 5 = 5 * 4fact
        num  = num * fact(num-1)
        return num
    
print(fact(5))
```

```python
# check palindrome
def palindrome(str):
    if len(str) <= 1:
        print("Palindrome") 
    else:
        if str[0] == str[-1]:
            palindrome(str[1:-1])
        else:
            print("Not Palindrome") 
```

```python
print(palindrome("madam"))
print(palindrome("abba"))
print(palindrome("rohit"))
```

```python
# fibonachi serise
def fib(m):
    if m==0 or m==1:
        return 1
    else:
        return fib(m-1) + fib(m-2)
    
print(fib(5))
```

#### Q.if 2 newborn rabbits are put in a pen, how many rabbits will be in the pen after 1 year?

```python
import time
def memo(m,d):
    if m in d:
        return d[m]
    else:
        d[m] = memo(m-1,d) + memo(m-2,d)
        return d[m]
    
start = time.time()
d = {0:1,1:1}
print(memo(48,d))
print(time.time()-start)
print(d)
```
