```python
def calculate_margin(revenue, expenses):
    profit = revenue - expenses
    margin = profit * 100/revenue
    return margin
    
```

```python
calculate_margin(50,10)
```

```python
#type hint
def calculate_profit_and_margin(revenue: float, expenses: float) -> float:
    profit = revenue - expenses
    margin = profit * 100/revenue
    return profit,margin
```

```python
calculate_profit_and_margin(50,10)
```

```python
response = calculate_profit_and_margin(50,10)
type(response)
```
