```python
# Strings are sequence of Unicode Character 
a = 'Rohit'
b = "Rohit"
c = """hello""" # use in multiline string
print(a)
print(b)
print(c)
```

#### Accessing Substrings from a String

```python
# concept of Indexing
c = "hello"
print(c)
```

```python
print(c[0])
```

```python
# type of indexing
# Positive Indexing - left to right (Start with 0)
# Negitive Indexing - right to left (Start with -1)
print(c[0]) 
print(c[-1])

```

```python
# Slicing
c = "Rohit Ravi"
print(c)
```

```python
# c[start : end]
# c[included : excluded]
print(c[0:4])
```

```python
print(c[6:])
```

```python
# c[start : end : steps]
print(c[0:5:2])
```

```python
print(c[-6:-1:2])
```

```python
print(c[::-1])
```
