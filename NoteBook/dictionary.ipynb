```python
# Rule
    # Dictionary has no indexing
    # it is a mutable types
    # key-> immutable, values-> they can be mutable 
    # Keys should be unique
```

```python
D = {"Name":"Rohit Ravi", "gender":"male"}
D
```

```python
D1 = {[1,2,3]:'Ravi'}
```
