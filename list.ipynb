```python
revenues = [50,60,55]
expenses = [10,30,35]
```

```python
for rev in revenues:
    # print("Reveue: ",rev)
    print(f"Revenue {rev}")
    
```

```python
for i in range(len(revenues)):
    profit = revenues[i] - expenses[i]
    margin = profit * 100 / revenues[i]
    # print(f"Profit {i}:{profit}" )
    print(f"Margin {i+1}: {margin:.2f}")
    
```

```python
margins = []

for i in range(len(revenues)):
    profit = revenues[i] - expenses[i]
    margin = profit * 100 / revenues[i]
    margins.append(margin)

margins
```

```python
financials = {
    "Q1": {"revenue":50, "expenses": 10},
    "Q2": {"revenue":60, "expenses": 30},
    "Q3": {"revenue":55, "expenses": 35},
    "Q4": {"revenue":-10, "expenses": 35},
}
for quarter, data in financials.items():
    if data["revenue"]< 0:
        print(f"Invalid revenue for {quarter}.Skipping.")
        continue
    margin = (data["revenue"] - data["expenses"]) / data["revenue"] * 100
    print(f"{quarter}: {margin:.2f}%")

```
